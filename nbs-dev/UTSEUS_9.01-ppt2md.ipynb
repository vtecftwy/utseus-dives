# PPT to Markdown Conversion

```python
from eccore.ipython import nb_setup
nb_setup()

from IPython.display import display, Markdown
from pathlib import Path
from docling.document_converter import DocumentConverter
from pypdf import PdfReader

import re

def printmd(text):
    display(Markdown(text))
```

```python
def pptx2md(fpaths, dst_dir=None):
    """Convert all pptx files in the fpaths into md"""
    if dst_dir is None: dst_dir = p2md_dir
    for ppt in fpaths:
        result = DocumentConverter().convert(ppt)
        md_path = dst_dir / f"{ppt.stem}.md"
        md_path.write_text(result.document.export_to_markdown(), encoding="utf-8")
        print(f"File converted: {md_path}")
        # break
```

## Clean up md files

```python
patterns = [
    r"2diyai.com",
    r"©\s*2020-2026\s*Etienne Charlier\s*\(DIY AI\)",
    r"©\s*2018-2026\s*Etienne Charlier\s*\(DIY AI\)\s*All Right Reserved",
    r"© 博舒华（上海）投资咨询有限公司 (DIY AI)",
    r"<!--\s*image\s*-->",
]

for p2md in p2md_dir.glob("*.md"):
    text = p2md.read_text(encoding="utf-8")
    for pattern in patterns:
        text = re.sub(pattern, "", text)
    # collapse runs of blank lines left behind
    text = re.sub(r"\n{3,}", "\n\n", text).strip() + "\n"
    p2md.write_text(text, encoding="utf-8")
    print(f"Cleaned: {p2md}")
```
