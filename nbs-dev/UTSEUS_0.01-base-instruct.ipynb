# Compare Base model and Instruct

```python
from eccore.ipython import nb_setup
from eccore.comms import clean_no_proxy_12vpn
nb_setup()
clean_no_proxy_12vpn()

from IPython.display import display, Markdown
from pathlib import Path
from docling.document_converter import DocumentConverter

from openai import OpenAI

import httpx

def printmd(text):
    display(Markdown(text))
```

## Setup LLM API with OpenAI

```python
BASE_URL = "http://localhost:9933/v1"

try:
    httpx.get(BASE_URL)
except httpx.RequestError as e:
    print(f"Check that llama-server is on and running. \nError connecting to {BASE_URL}: {e}")
    raise RuntimeError(f"Failed to connect to {BASE_URL}") from e

client = OpenAI(
    base_url=BASE_URL,
    api_key="no-key-required",
    timeout=300.0,
    max_retries=0,
)
printmd(f"### Available Models:")
for m in client.models.list():
    printmd(f"- `{m.id}`")
```

```python
model = "llama3.2-base"
```

```python
prompt = (
    "What is the capital city of France?"
)

# Replace the second alias with the one configured in your server.
for model_name in ["llama3.2-base", "llama3.2-instruct"]:
    response = client.completions.create(
        model=model_name,
        prompt=prompt,
        max_tokens=3000,
        temperature=0.7,
        stream=False,
        seed=888
    )

    print(f"\n===== {model_name}: raw completion =====")
    printmd(response.choices[0].text)
```

### Using `client.responses.create`

openai documentation for [`create`](https://developers.openai.com/api/reference/python/resources/responses/methods/create)

```python
response = client.responses.create(
    model=model,
    input="Explain APIs in one sentence",
    temperature=0.7,
    # max_tokens=10_000,
    reasoning={'effort': 'none'},
    extra_body={
    "chat_template_kwargs": {
        "enable_thinking": False
        },
    "reasoning_format": "deepseek",
    }
)
```

```python
printmd(response.output[0].content[0].text)
printmd(f"object: `{response.object}`")
```

```python
for attr in response.model_fields_set:
    try:
        if isinstance(getattr(response, attr), str):
            length = ""
        else:
            length = "*" +str(len(getattr(response, attr))) + " elements*: "
    except TypeError:
        length = ""
    printmd(f"- `{attr}`:&nbsp;&nbsp;&nbsp;  {length}{getattr(response, attr)}")
```

```python
for output in response.output:
    for attr in output.model_fields_set:
        print(f"- `{attr}`: {getattr(output, attr)}")
    printmd("Printing message:")
    printmd(f"#### `content`:")
    for content in output.content:
        printmd(f"type: `{content.type}`")
        printmd(content.text)
    printmd("---")
```
